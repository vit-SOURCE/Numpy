```python
#suppression des colonnes inutiles 
df = df.drop(columns=['CustomerID', 'Count', 'Country', 'State', 'City', 
                       'Zip Code', 'Lat Long', 'Latitude', 'Longitude', 
                       'Gender', 'Churn Score', 'Churn Reason', 'Churn Label'])
```

```python
df
```

```python
#conversion des valeurs string en float dans la colonne Total Charges 
df["Total Charges"]=pd.to_numeric(df["Total Charges"], errors="coerce")
df.isnull().sum()
df['Total Charges'] = df['Total Charges'].fillna(0)
df.isnull().sum()
```

```python
#division du dataframe (split)
from sklearn.model_selection import train_test_split
Y=df["Churn Value"]
X=df.drop(columns="Churn Value")
X_train,X_test,Y_train,Y_test=train_test_split(  X, Y,
    test_size=0.2,      
    random_state=42,     
    stratify=Y)
X_test.shape
X_train.nunique()
```

```python

for col in ['Paperless Billing', 'Online Security', 'Online Backup', 
            'Device Protection', 'Tech Support', 'Streaming TV', 
            'Streaming Movies', 'Multiple Lines']:
    print(col, X_train[col].unique())
    
```

```python
#classification des colonnes par categories
list_bi=["Senior Citizen",	"Partner",	"Dependents","Phone Service","Paperless Billing"]
list_cat=["Multiple Lines","Online Backup",	"Device Protection","Tech Support",	"Streaming TV","Streaming Movies","Internet Service","Contract","Payment Method","Online Security"]
list_num=["Tenure Months","Monthly Charges","Total Charges","CLTV"]
#importation des bibliothèques
from sklearn.preprocessing import OrdinalEncoder,OneHotEncoder,StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix

#instanciations des encoders et du scaler
ordinal=OrdinalEncoder()
Hot=OneHotEncoder()
num= StandardScaler()

#mise en place du  ColumnTransformer
preprocessor=ColumnTransformer(
    transformers=
    [('binaire',ordinal,list_bi),
     ('multi',Hot,list_cat),
     ('nombre',num,list_num)
    ]
    
)


X_train_processed = preprocessor.fit_transform(X_train) # fonction de transformation appliquée sur les données d'entrainements
X_test_processed = preprocessor.transform(X_test) # fonction de transformation appliquée sur les données de test mais sans le fit


model=LogisticRegression() #instanciation de l'estimator
# Le modèle apprend la relation entre les caractéristiques (X_train_processed) 
# et le churn réel (Y_train), en ajustant ses coefficients internes
model.fit(X_train_processed,Y_train) 

#le modèle fait sa première prédiction avec les données test
predictions = model.predict(X_test_processed)
predictions[:10]

# score
accuracy_score(Y_test,predictions)
#vérification de la précision
confusion_matrix(Y_test, predictions)
```
